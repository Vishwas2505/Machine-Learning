```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
x=np.arange(1,21).reshape(-1,1)
y=np.array([10,12,15,18,20,25,30,28,35,40,45,42,50,55,60,58,65,70,72,75])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
reg=DecisionTreeRegressor(random_state=42)
reg.fit(x_train,y_train)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf=RandomForestRegressor(n_estimators=100,random_state=42)
rf.fit(x_train,y_train)
```

```python
x_grid=np.linspace(1,20,200).reshape(-1,1)
plt.scatter(x,y,color='black')
y_pred=reg.predict(x_grid)

plt.plot(x_grid,y_pred,color='red',label='decisiotree')
plt.plot(x_grid,rf.predict(x_grid),color='green',label='RandomForestRegressor')
plt.xlabel("House size index")
plt.ylabel("price")
plt.title("decision tree regression -salary predection")
plt.show()
```

```python
from sklearn.metrics import *
dt_pred=reg.predict(x_test)
rf_pred=rf.predict(x_test)

print("Decision Tree Regression")
print("MSE:",mean_squared_error(y_test,dt_pred))
print("R2:",r2_score(y_test,dt_pred))

print("Random forest Regression")
print("MSE:",mean_squared_error(y_test,rf_pred))
print("R2:",r2_score(y_test,rf_pred))
```
