```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
x=np.array([50,120,180,250,320,380]).reshape(-1,1)
y=np.array([500,1000,1500,1800,2500,2500])
```

```python
from sklearn.tree import DecisionTreeRegressor
reg=DecisionTreeRegressor(max_depth=3,random_state=42)
reg.fit(x,y)
```

```python
x_grid=np.linspace(0,400,200).reshape(-1,1)
y_pred=reg.predict(x_grid)

plt.scatter(x,y,color='blue')
plt.plot(x_grid,y_pred,color='red')
plt.xlabel("Units consumed")
plt.ylabel("electicity bill")
plt.title("decision tree regression -Electicity bill slabs")
plt.show()
```

```python
units=float(input("Enter electricity units"))
bill=reg.predict([[units]])

print(f"predicited electricity bill: ${bill[0]:.0f}")
```

# predict salary based on experience

```python
x1=np.array([1,2,3,4,5,7,8,9]).reshape(-1,1)
y1=np.array([3,3.5,4,4.2,6,6.5,7,7.2])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x1,y1,test_size=0.25,random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
reg=DecisionTreeRegressor(max_depth=3,random_state=42)
reg.fit(x_train,y_train)
```

```python
x_grid=np.linspace(1,8,100).reshape(-1,1)
y_pred=reg.predict(x_grid)

plt.scatter(x1,y1,color='blue')
plt.plot(x_grid,y_pred,color='red')
plt.xlabel("years of experince")
plt.ylabel("salary")
plt.title("decision tree regression -salary predection")
plt.show()
```

```python
dt_deep=DecisionTreeRegressor(max_depth=None)
dt_deep.fit(x_train,y_train)

y_pred_deep=dt_deep.predict(x_grid)

plt.scatter(x1,y1)
plt.plot(x_grid,y_pred_deep,color='green')
plt.title("over tree")
plt.show()
```

```python
dt_con=DecisionTreeRegressor(max_depth=3)
dt_con.fit(x_train,y_train)

y_pred_con=dt_con.predict(x_grid)

plt.scatter(x1,y1)
plt.plot(x_grid,y_pred_con,color='orange')
plt.title("controlled fit")
plt.show()
```

```python
from sklearn.metrics import *

y_test_pred_deep=dt_deep.predict(x_test)
y_test_pred_controlled=dt_con.predict(x_test)

print("deep tree metrics")
print("MSE:",mean_squared_error(y_test,y_test_pred_deep))
print("r2",r2_score(y_test,y_test_pred_deep))

print("Controlled tree metrics")
print("MSE:",mean_squared_error(y_test,y_test_pred_controlled))
print("r2",r2_score(y_test,y_test_pred_controlled))
```
